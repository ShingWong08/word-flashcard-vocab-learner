### Card deck

Creating a new deck

```python
path = cardDeck.defaultDeckPath()

print(path)

cards = cardDeck.loadDeck(path)             # Try to load a deck from default path
```

Adding new word(s) to the card deck

```python
word = "apple"

result = freeDict.getWordDefinition(word)

newCard = Card.fromLookup(result)           # Creates a new card
  
added = cardDeck.addCard(cards, newCard)    
# Saves the new card to the deck. Saves nothing if the card already exists. (i.e. The deck contains the same word card)                                          
# added is True if the deck does not contain the word card, vice versa.

print(f"Not in deck?: {added}")

cardDeck.backupDeck(path)                   
cardDeck.saveDeck(cards, path)              # Double saving of the updated deck
```

Load word cards from the deck

```python
loadedCards = cardDeck.loadDeck(path)

for card in cardDeck.sortByWord(loadedCards):
    print(freeDict.formatWordDefinition(card.toDict()))
    print()

dueToday = cardDeck.getDueCards(loadedCards, date.today())
print(f"Cards due today: {len(dueToday)}")
```
